```python



# ============================================================
# 1. ADD customer_unique_id TO MAIN DATAFRAME
# ============================================================

# Only run this merge if customer_unique_id is not already present
if "customer_unique_id" not in df.columns:

    df = df.merge(
        customers[
            ["customer_id", "customer_unique_id"]
        ],
        on="customer_id",
        how="left"
    )


# ============================================================
# 2. PREPARE DATE
# ============================================================

df["day"] = pd.to_datetime(df["order_approved"]).dt.normalize()


# ============================================================
# 3. FIND FIRST PURCHASE DAY OF EVERY CUSTOMER
# ============================================================

first_customer_day = (
    df.groupby("customer_unique_id")["day"]
      .min()
)

df["customer_first_day"] = (
    df["customer_unique_id"]
    .map(first_customer_day)
)


# ============================================================
# 4. FIND FIRST ACTIVE DAY OF EVERY SELLER
# ============================================================

first_seller_day = (
    df.groupby("seller_id")["day"]
      .min()
)

df["seller_first_day"] = (
    df["seller_id"]
    .map(first_seller_day)
)


# ============================================================
# 5. BASIC DAILY KPIs
# ============================================================

daily_df = (
    df.groupby("day")
      .agg(

          # Revenue / GMV
          daily_gmv=("price", "sum"),

          # Total unique orders
          daily_orders=("order_id", "nunique"),

          # Actual unique customers
          daily_customers=("customer_unique_id", "nunique"),

          # Number of items sold
          units_sold=("order_item_id", "count"),

          # Sellers who actually sold something that day
          active_sellers=("seller_id", "nunique"),

          # Average price of an individual item
          average_item_price=("price", "mean")
      )
      .reset_index()
)


# ============================================================
# 6. CUSTOMER / ORDER DRIVER METRICS
# ============================================================

# Orders per customer
daily_df["orders_per_customer"] = (
    daily_df["daily_orders"]
    / daily_df["daily_customers"]
)


# Average Order Value
daily_df["average_order_value"] = (
    daily_df["daily_gmv"]
    / daily_df["daily_orders"]
)


# Items per order
daily_df["items_per_order"] = (
    daily_df["units_sold"]
    / daily_df["daily_orders"]
)


# Average customer spending
daily_df["avg_spend_per_customer"] = (
    daily_df["daily_gmv"]
    / daily_df["daily_customers"]
)


# ============================================================
# 7. NEW CUSTOMERS
# ============================================================

new_customers = (
    df[
        df["day"] == df["customer_first_day"]
    ]
    .groupby("day")["customer_unique_id"]
    .nunique()
    .rename("new_customers")
)


daily_df = daily_df.merge(
    new_customers,
    on="day",
    how="left"
)


# ============================================================
# 8. RETURNING CUSTOMERS
# ============================================================

returning_customers = (
    df[
        df["day"] > df["customer_first_day"]
    ]
    .groupby("day")["customer_unique_id"]
    .nunique()
    .rename("returning_customers")
)


daily_df = daily_df.merge(
    returning_customers,
    on="day",
    how="left"
)


# Replace missing values with 0
daily_df["new_customers"] = (
    daily_df["new_customers"]
    .fillna(0)
    .astype(int)
)

daily_df["returning_customers"] = (
    daily_df["returning_customers"]
    .fillna(0)
    .astype(int)
)


# ============================================================
# 9. NEW SELLERS
# ============================================================

new_sellers = (
    df[
        df["day"] == df["seller_first_day"]
    ]
    .groupby("day")["seller_id"]
    .nunique()
    .rename("new_sellers")
)


daily_df = daily_df.merge(
    new_sellers,
    on="day",
    how="left"
)


daily_df["new_sellers"] = (
    daily_df["new_sellers"]
    .fillna(0)
    .astype(int)
)


# ============================================================
# 10. TOTAL SELLER COUNT
# ============================================================

# seller_count = total seller base accumulated up to that day

daily_df = daily_df.sort_values("day")

daily_df["seller_count"] = (
    daily_df["new_sellers"]
    .cumsum()
)


# ============================================================
# 11. SELLER ORDERS
# ============================================================

# Important:
# One order can contain products from multiple sellers.
# Therefore count unique order-seller combinations.

seller_orders = (
    df[
        ["day", "seller_id", "order_id"]
    ]
    .drop_duplicates()
    .groupby("day")
    .size()
    .rename("seller_orders")
)


daily_df = daily_df.merge(
    seller_orders,
    on="day",
    how="left"
)


# ============================================================
# 12. ORDERS PER ACTIVE SELLER
# ============================================================

daily_df["orders_per_seller"] = (
    daily_df["seller_orders"]
    / daily_df["active_sellers"]
)


# ============================================================
# 13. UNITS SOLD PER ACTIVE SELLER
# ============================================================

daily_df["units_sold_per_seller"] = (
    daily_df["units_sold"]
    / daily_df["active_sellers"]
)


# ============================================================
# 14. GMV / REVENUE PER ACTIVE SELLER
# ============================================================

daily_df["gmv_per_active_seller"] = (
    daily_df["daily_gmv"]
    / daily_df["active_sellers"]
)


# ============================================================
# 15. SORT DATA BY DAY
# ============================================================

daily_df = (
    daily_df
    .sort_values("day")
    .reset_index(drop=True)
)


# ============================================================
# 16. CREATE DAY INDEX: 1, 2, 3, 4 ...
# ============================================================

daily_df["day_index"] = range(
    1,
    len(daily_df) + 1
)


# Make day_index actual dataframe index
daily_df = daily_df.set_index("day_index")


# ============================================================
# 17. FINAL COLUMN ORDER
# ============================================================

daily_df = daily_df[[
    
    "day",

    # Main KPI
    "daily_gmv",

    # Customer metrics
    "daily_customers",
    "new_customers",
    "returning_customers",
    "orders_per_customer",
    "avg_spend_per_customer",

    # Order metrics
    "daily_orders",
    "average_order_value",
    "items_per_order",

    # Product metrics
    "units_sold",
    "average_item_price",

    # Seller metrics
    "seller_count",
    "new_sellers",
    "active_sellers",
    "seller_orders",
    "orders_per_seller",
    "units_sold_per_seller",
    "gmv_per_active_seller"
]]


# ============================================================
# 18. DISPLAY
# ============================================================

daily_df.head()
```
